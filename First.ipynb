### pandas 라이브러리 가져오기 & CSV 파일 읽기 & 파일 전체 출력해보기

```python
import pandas as pd

df = pd.read_csv('성적.csv') # df는 dataframe의 약자, 2차원 표 형태의 자료구조로, 행과 열로 이루어져 있음
df
```

### 시리즈 (1차원 데이터)
대괄호 한번

**df가 아니라 df1, 2 ...으로 변수를 따로 정의하는 이유: 원본 데이터를 건드리지 않고 여러 가지를 해보기 위함**

```python
df1 = df['이름'] # series
df1
```

### 데이터프레임 (2차원 데이터, 표)
대괄호 2개

```python
df2 = df[['이름']] # dataframe
df2
```

### Cols (Columns)
- 원하는 부분 (여기서는 이름, z점수, 성취도)만 출력하기

```python
cols = ['이름', 'z점수', '성취도']
df3 = df[cols]
df3
```

### df[df['조건']]
- 조건에 맞는 행 (row)만 골라냄
- 예시에서는 z점수가 0보다 작은 행만 출력

```python
df4 = df[df['z점수'] < 0]
df4
```

### 뒤에 항목 추가
- 기본 조건 필터링은 조건에 맞는 행 전체를 출력, 아래처럼 하면 해당하는 행들 중 원하는 부분 (여기서는 이름)만 출력

```python
df5 = df[df['z점수'] > 0]['이름']
df5
```

### 뒤쪽을 시리즈가 아닌 데이터프레임으로 설정

```python
df6 = df[df['z점수'] > 0][['이름', 'z점수', '성취도']]
df6
```

### 슬라이싱
- 지정한 행 또는 구간만 출력

```python
df7 = df[0:10]
df7
```
