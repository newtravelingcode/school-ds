## 중복값 처리

### duplicated()
- 행마다 중복 여부 판정, True 또는 False로 나옴

```python
df.duplicated()
```

### duplicated().sum()
- sum()을 추가 -> 개수로 출력됨

```python
df.duplicated().sum()
```

### drop.duplicates
- 중복값 삭제 (drop)
- 이후 다시 duplicated().sum()을 실행해 중복값이 사라졌는지 확인 -> 사라짐 (0개)

```python
df = df.drop_duplicates()
df.duplicated().sum()
```

### subset
- subset을 이용해 특정 열에서의 중복 판정 가능
- print를 사용하지 않으면 마지막 줄만 출력됨 - 원래 정보를 출력하는 코드는 print를 사용하는 것이 적절, 이 코드들에서도 정보 출력시키는 줄들을 print()에 집어넣어도 정상 작동함
- 마찬가지로 그냥 쓰면 모든 행에 대해 True와 False로 나열, sum을 붙이면 개수만

### subset을 사용하면 방금 전에 중복값을 지웠는데 중복값이 또 나오는 이유
- 일반 duplicates는 모든 열이 **완전히 동일한** 행들만 파악함
- 따라서 일부분만 같은 행 (여기서는 연도)들은 subset을 사용해야 나타남

```python
print(df.duplicated(subset=['년']))
df.duplicated(subset=['년']).sum()
```

### subset에서 중복값 삭제
- 기본적으로 동일하지만 subset만 지정해준다고 보면 됨
- 삭제 후 개수 출력해서 확인하기

```python
df = df.drop_duplicates(subset=['년'])
df.duplicated(subset=['년']).sum()
```

## 결측치 처리

### isna()
- is N/A 라는 뜻으로, sum을 붙여 개수로 파악

```python
df.isna().sum()
```

### 결측치 삭제
- 여기서부터는 원본 데이터 유지를 위해 예시마다 변수 새로 만들기 (df2, df3, ...)
- 중복값과 유사하게 dropna 사용
- 이후 개수 다시 파악하기

```python
df2 = df.dropna()
df2.isna().sum()
```

### subset 지정해서 결측치 제거하기

```python
df3 = df.dropna(subset=['평균기온(°C)'])
df3.isna().sum()
```

### 지정한 조건에 맞는 부분 선택 & 결측치 파악
- 아래 예시는 평균기온이 11 이상이면서 평균기온이 결측치가 아닌 행만 선택됨
- 이후 그 부분에서의 결측치 수 파악하기

```python
df4 = df[(df['평균기온(°C)']>11) & (df['평균기온(°C)'].notnull())]
df4.isna().sum()
```

### 평균값 구하기 (mean)
- 아래부터 예시는 모두 평균기온에 대해서 조사

```python
df['평균기온(°C)'].mean()
```

### 중앙값 구하기 (median)

```python
df['평균기온(°C)'].median()
```

### 최빈값 구하기 (mode)
- mode라는 단어가 잘 와닿지 않으니 주의해서 암기

```python
df['평균기온(°C)'].mode()[0]
```

### 결측치를 평균값으로 대체해서 채우기

```python
df['평균기온(°C)'] = df['평균기온(°C)'].fillna(df['평균기온(°C)'].mean())
df
```
